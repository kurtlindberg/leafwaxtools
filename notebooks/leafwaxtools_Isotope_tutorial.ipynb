# leafwaxtools Tutorial: Isotope API class

Import the leafwaxtools Chain module/class and other necessary packages for this tutorial. It's recommended to use the leafwaxtools conda environment (environment.yml) which contains all of the packages you'll need for this notebook.

```python
from leafwaxtools import Isotope

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Importing Leaf Wax Data

```python
cf8sed_data_df = pd.read_csv(r'../leafwaxtools/data/Thomas_CF8_2023.csv')

cf8sed_data_df
```

## Preparing Leaf Wax Stable Isotope Data

```python
cf8sed_nacid_d2h_df = cf8sed_data_df[
    [
        'd2HC20',
        'd2HC22',
        'd2HC24',
        'd2HC26',
        'd2HC28',
        'd2HC30',
    ]
]

cf8sed_nacid_d2h_df
```

## Isotope Value Range: Isotope.value_range()

```python
cf8sed_nacid_d2h_value_range = Isotope(cf8sed_nacid_d2h_df).value_range()

fig, ax = plt.subplots()

ax.plot(
    cf8sed_data_df.ageMedian, cf8sed_nacid_d2h_value_range, color='black'
)
ax.set_xlabel("Age (cal yr BP)")
ax.set_ylabel(u"$\u03B4^{2}$H Range ($C_{20}$-$C_{30}$; $\u2030$ VSMOW)")

plt.show()
```

## Chain-Length Concentration-Weighted Average Isotope Value: Isotope.concentration_avg()

```python
fig, ax = plt.subplots()

ax.plot()

plt.show()
```

## Calculate Epsilon: Isotope.epsilon()

```python
fig, ax = plt.subplots()

ax.plot()

plt.show()
```

## Calculate Isotope Values Using Epsilon: Isotope.wax_to_source()

```python
fig, ax = plt.subplots()

ax.plot()

plt.show()
```

## Correlation Matrices: Chain.correlation_rvals() & Chain.correlation_pvals()

```python
cf8sed_nacid_d2h_correlation_rvals = Isotope(cf8sed_nacid_d2h_df).correlation_rvals()
cf8sed_nacid_d2h_correlation_pvals = Isotope(cf8sed_nacid_d2h_df).correlation_pvals()


fig, axs = plt.subplots(2,2,layout='constrained')

corr_mask = np.triu(np.ones_like(cf8sed_nacid_d2h_correlation_rvals, dtype=bool))

ax = axs[0,0]
ax.set_title("Pearson r-values")
sns.heatmap(
    ax=ax, data=cf8sed_nacid_d2h_correlation_rvals,
    cmap='Reds', mask=corr_mask, annot=True, fmt='.2f', vmin=0, vmax=1,
    linewidths=0, linecolor='black', clip_on=False,
    cbar=True, cbar_kws={'label': "r", 'shrink': 0.66, 'location': 'bottom'},
    xticklabels=["C20","C22","C24","C26","C28","C30"],
    yticklabels=["C20","C22","C24","C26","C28","C30"]
)

ax = axs[0,1]
ax.set_title("Pearson p-values")
sns.heatmap(
    ax=ax, data=cf8sed_nacid_d2h_correlation_pvals,
    cmap='Blues', mask=corr_mask, annot=True, fmt='.2f',
    linewidths=0, linecolor='black', clip_on=False,
    cbar=True, cbar_kws={'label': "p", 'shrink': 0.66, 'location': 'bottom'},
    xticklabels=["C20","C22","C24","C26","C28","C30"],
    yticklabels=["C20","C22","C24","C26","C28","C30"]
)

fig.delaxes(axs[1,0])
fig.delaxes(axs[1,1])

plt.show()
```
